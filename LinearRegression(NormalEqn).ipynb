```python
import numpy as np
import pandas as pd
```

```python
from sklearn.datasets import fetch_california_housing

data = fetch_california_housing(as_frame=True)

df = data.frame

df.head()
```

```python
X = df.drop('MedHouseVal', axis = 1)
y = df['MedHouseVal']
```

```python
Xmat = np.c_[np.ones(X.shape[0]), X]
ymat = np.array(y.values)
```

```python
def fit(Xmat, ymat):
  theta = np.linalg.inv(Xmat.T @ Xmat) @ (Xmat.T @ ymat)
  return theta
```

```python
theta = fit(Xmat, ymat)
```

```python
b = theta[0]
w = theta[1:]
```

```python
def predict(Xmat, theta):
  return Xmat @ theta
```

```python
y_pred = predict(Xmat, theta)
```

```python
from sklearn.metrics import root_mean_squared_error, r2_score
```

```python
print("RMSE : ", root_mean_squared_error(y, y_pred))
print('R2 Score : ',r2_score(y, y_pred))
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(X, y)

print("Scratch:", w, b)
print("Sklearn:", model.coef_, model.intercept_)
```
